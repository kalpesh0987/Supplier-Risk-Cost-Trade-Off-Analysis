# Supplier Risk & Cost Trade-Off Analysis

This notebook analyzes suppliers based on **cost, risk, reliability, and sustainability** to identify optimal sourcing strategies.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset

```
suppliers = pd.read_csv('suppliers_risk.csv')
suppliers
```

## Data Exploration

```
print(suppliers.describe())
print('\nAverage Cost:', suppliers['UnitCost'].mean())
print('Average Risk:', suppliers['RiskScore'].mean())
```

## Risk vs. Cost Scatterplot

```
plt.figure(figsize=(7,5))
sns.scatterplot(data=suppliers, x='UnitCost', y='RiskScore', hue='Country', s=100)
plt.axvline(suppliers['UnitCost'].mean(), color='red', linestyle='--', label='Avg Cost')
plt.axhline(suppliers['RiskScore'].mean(), color='blue', linestyle='--', label='Avg Risk')
plt.title('Supplier Risk vs. Cost')
plt.xlabel('Unit Cost')
plt.ylabel('Risk Score')
plt.legend()
plt.show()
```

## Supplier Scoring Model

```
# Higher score = better supplier
suppliers['Score'] = (0.6 * (1/suppliers['UnitCost'])) + (0.3 * (1/suppliers['RiskScore'])) + (0.1 * (suppliers['Sustainability']/100))
suppliers = suppliers.sort_values(by='Score', ascending=False)
suppliers[['Supplier','UnitCost','RiskScore','Reliability','Sustainability','Score']]
```

## Reliability vs. Sustainability

```
plt.figure(figsize=(7,5))
sns.barplot(data=suppliers, x='Supplier', y='Reliability', color='skyblue', label='Reliability')
sns.barplot(data=suppliers, x='Supplier', y='Sustainability', color='lightgreen', alpha=0.7, label='Sustainability')
plt.title('Supplier Reliability vs Sustainability')
plt.ylabel('Percentage')
plt.legend()
plt.show()
```
